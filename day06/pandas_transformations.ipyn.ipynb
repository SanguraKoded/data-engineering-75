```python
path = r'C:\Users\H\Downloads\data-engineering-75\day07\weather_data_5_cities.json'

# Read it directly with pandas using lines=True
df = pd.read_json(path, lines=True)

# Flatten the nested dictionaries (like coord, main, wind) into separate columns
df_flat = pd.json_normalize(df.to_dict(orient='records'))

df_flat.head()
```

```python
df['temperature'] = df['main'].apply(lambda x : x['temp'])
```

```python
df['humidity'] = df['main'].apply(lambda x: x['humidity'])
```

```python
df.groupby('name')[['temperature', 'humidity']].mean()
```

```python
df['temp_category'] = df['temperature'].apply(lambda x: 'hot' if x > 300 else 'warm' if x > 290 else 'cold')
```

```python
df2 = df.sort_values(by='temperature', ascending = False).reset_index(drop=True)
```
